```python
import pandas as pd
from matplotlib import pyplot as plt 
import seaborn as sns 
```

```python
covid_country_data=pd.read_csv("country_wise_latest.csv")
covid_country_data.head()
```

```python
top10 = covid_country_data.sort_values(by='Confirmed', ascending=False).head(10)

plt.figure(figsize=(10, 6))
sns.barplot(y='Country/Region', x='Confirmed', data=top10)
plt.title('Top 10 Countries by Confirmed COVID-19 Cases')
plt.tight_layout()
plt.show()
```

```python
top10 = covid_country_data.sort_values(by='Confirmed', ascending=False).tail(30)

plt.figure(figsize=(10, 6))
sns.barplot(y='Country/Region', x='Confirmed', data=top10)
plt.title('Top 10 Countries by Lowest Confirmed COVID-19 Cases')
plt.tight_layout()
plt.show()
```

```python
region_data = covid_country_data.groupby('WHO Region')['Confirmed'].sum().reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(x='WHO Region', y='Confirmed', data=region_data)
plt.xticks(rotation=30)
plt.title('Confirmed COVID-19 Cases by WHO Region')
plt.tight_layout()
plt.show()
```

```python
top_10_active_cases=covid_country_data.sort_values(by="Active",ascending=False).head(10)

sns.barplot(data=top_10_active_cases,y="Country/Region",x="Active")
plt.title('Top 10 Countries High Active Cases')
plt.xlabel('Active')
plt.ylabel('Country/Region')
plt.tight_layout()
plt.show()
```

```python
covid_country_data.head(2)
```

```python
top10_death_rate = covid_country_data.sort_values(
    by='Deaths / 100 Cases', ascending=False
).head(10)

plt.figure(figsize=(10, 6))
sns.barplot(
    y='Country/Region',
    x='Deaths / 100 Cases',
    data=top10_death_rate
)
plt.title('Top 10 Countries by Death Rate (per 100 Cases)')
plt.xlabel('Deaths per 100 Cases')
plt.ylabel('Country/Region')
plt.tight_layout()
plt.show()
```

```python
top10_death_lowest_rate = covid_country_data.sort_values(
    by='Deaths / 100 Cases', ascending=True
).head(30)

plt.figure(figsize=(10, 6))
sns.barplot(
    y='Country/Region',
    x='Deaths / 100 Cases',
    data=top10_death_lowest_rate
)
plt.title('Top 30 Countries by Lowest Death Rate (per 100 Cases)')
plt.xlabel('Deaths per 100 Cases')
plt.ylabel('Country/Region')
plt.tight_layout()
plt.show()
```

```python
top10_weekly_increase = covid_country_data.sort_values(
    by='1 week change', ascending=False
).head(10)

plt.figure(figsize=(10, 6))
sns.barplot(
    y='Country/Region',
    x='1 week change',
    data=top10_weekly_increase
)
plt.title('Top 10 Countries by Weekly Increase in COVID-19 Cases')
plt.xlabel('1 week change')
plt.ylabel('Country/Region')
plt.tight_layout()
plt.show()
```
