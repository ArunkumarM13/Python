## Pull up the Average Insurance Charges Based on region

```python
insurance_date.groupby(by="region")["charges"].mean().round().sort_values() #Chain of Methods
```

## Pull up the Average Insurance Charges Based on region and Sex

## Discrete and Continous Date

* Discrete - Just Countable/Can't be segmented

* continous - Measurable /Have Units/Can also be segmented

```python
insurance_date.groupby(by=["region","sex"])["charges"].mean().round().sort_values(ascending=False)
```

# Pivot Table

```python
pd.pivot_table(data=insurance_date,values="charges",index="region",columns=["sex"]).round() # Presentable comparing 2 categorical features vs 1 continous
```

## Find out the average insurance charges based on region.

```python
insurance_date.groupby(by=["region","sex"])["charges"].agg(func=["mean","min","max","sum"]).round()
```

## Create a new column in the same dateframe with the name "Eligibilty Criteria" where if age is > 20 , say Eligibilty else Not.

```python
insurance_date[insurance_date["age"] > 20]
```

```python
def get_age(x):
    if x > 20:
        return "Eligibile"
    else:
        return "Not Eligibile"
```

```python
insurance_date["Eligibilty Criteria"]=insurance_date["age"].apply(func=get_age)
insurance_date
```

# Lambda Function

* Anonymous Function/Nameless Function

```python
insurance_date["EC"]=insurance_date["age"].apply(lambda x: "Eligibile" if x>20 else "Not Eligibile")
insurance_date
```
