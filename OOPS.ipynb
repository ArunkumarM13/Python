## Real world analogy 1 :

### I want to reuse the same function for team members and mangers, but they have small differences. How do I do that ?

Both Team members and Managers should be able to log in , but:
* **Team members** - should log in, view tasks , and submit tasks.
* **Managers** - should log in, view tasks, and assign tasks.

```python
def login():
    print(f"{user_data["name"]} has logged in")
    
def view_tasks(user_data):
    print(f"{user_data["name"]} has viewed the tasks")
    
def submit_tasks(user_data):
    if user_data["role"] == "team_member":
        print(f"{user_data["name"]} has submitted the tasks ")

    else:
        print("Managers no need to submit the tasks")

    
def assign_tasks(user_data):
    if user_data["role"] == "manager":
        print(f"{user_data["name"]} has assigned the tasks ")

    else:
        print("Team members can't assign the tasks")

    
# Data collection
user_data={"name":"arun","role":"team_member"}

# Functions to be called 
login()
view_tasks(user_data)
submit_tasks(user_data)
assign_tasks(user_data)
```

## With OOPS

* Class - Blueprint that contains attributes of a particular object and it's all the functions.
* Object - Instance of a class

```python
class Employee:
    def __init__(self,name):
        self.name = name

    def login(self):
        print(f"{self.name} has logged in")
    
    def view_tasks(self):
        print(f"{self.name} has viewed the tasks")

class TeamMember(Employee):
    def submit_tasks(self):
        print(f"{self.name} has submitted the tasks ")


class Manager(Employee):
    def assign_tasks(self):
        print(f"{self.name} has assigned the tasks ")
```

```python
Arun=TeamMember("Arun")
John=Manager("John")
```

```python
Arun.login()
Arun.submit_tasks()
Arun.view_tasks()
```

```python
John.assign_tasks()
```
