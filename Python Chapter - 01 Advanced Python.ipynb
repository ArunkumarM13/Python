### Positional Arguments

```python
def get_details(name,place,expected_price):
    print("Client Name:",name)
    print("Place:",place)
    print("Expected Price:",expected_price)
```

```python
get_details(name="Arunkumar",place="Chennai",expected_price="30L")
```

### Default Parmeters & Arguments

```python
def get_details_Chennai(name,expected_price,place="Chennai"):
    print("Client Name:",name)
    print("Place:",place)
    print("Expected Price:",expected_price)
```

```python
get_details_Chennai('Arunkumar',"50L")
```

```python
get_details_Chennai('Arunkumar',"50L","Bangalore")
```

### Variable - length arguments

```python
def get_details_Chennai(name,place,*expected_price): # * - Used for multiple answer from the user
    print("Client Name:",name)
    print("Place:",place)
    print("Expected Price:",expected_price)
```

```python
get_details_Chennai('Arunkumar',"Bangalore",60,56,45,90)
```

### Keyword Arguments 

```python
def get_details_Chennai(name,place,**expected_price): # ** - Used for multiple answer specified from the user
    print("Client Name:",name)
    print("Place:",place)
    print("Expected Price:",expected_price)
```

```python
get_details_Chennai("Arunkumar","Bangalore",ask_1=60,ask_2=56,ask_3=45,ask_4=90)
```

## 5.4 Challenge - 2 Find the Factors of a Number 

If the user enters 20, output should be "The factors of 20 is 1,2,4,5,10,20"

```python
user_input=int(input("Enter the Number"))

for i in range(1,user_input+1):
    if user_input%i==0:
        print(i)
    
```

```python
user_input=int(input("Enter the Number"))

result = []

for i in range(1,user_input+1):
    if user_input%i==0:
        result.append(i)
print("The factors of {} is {}" .format(user_input,result))
```
