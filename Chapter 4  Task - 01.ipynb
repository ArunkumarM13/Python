# 2.Create a list of numbers and print only even numbers using a for loop.

```python
A=range(1,31)
even_numbers=[]

for i in A:
    if i%2 == 0:
        even_numbers.append(i)

even_numbers
```

# 3.Write a program that handles division by zero using try-except.

```python
numerator = int(input("Enter the Numerator"))
den = int(input("Enter the denominator"))

try:
    results = numerator/den
    print(results)

except ZeroDivisionError:
    print("Invaild Entry ! Enter the number other than 0")

print("Program Ended")
```

# 4.Given a string, count how many vowels it has.

```python
vowels =("a","e","i","o","u")
count = 0

user_input=input("Enter your Name").lower()
for i in user_input:
    if i in vowels:
        count=count+1
print("Total numbers of Vowels are:",count)
```

# 5.Reverse a list without using built-in functions.

```python
original_list=[1,2,3,4,5]
updated_list=[]

i=len(original_list)-1

while i>=0:
    updated_list.append(original_list[i])
    i = i-1

updated_list
```
