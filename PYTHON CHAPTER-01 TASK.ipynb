## Rock, Paper, Scissors Game

```python
Game_character=("Rock","Paper","Scissors")
import random
print("Welcome to Rock, Paper, Scissors Game")

random_1=random.choice(Game_character)

while True:
    user_input=input("Enter the Guess: Rock ,Paper,Scissors")
    if user_input not in Game_character:
        print("Invaild Character,Please enter your guess of Rock,Paper,Scissors")
    elif user_input == random_1:
        print("It's a Tie")
    elif (user_input == "Rock"and random_1 =="Scissors")or(user_input == "Scissors"and random_1 == "Paper")or(user_input =="Paper"and random_1 =="Rock"):
        print("You Win")
    else:
        print("Computer Wins")
    while True:
        play_again=input("Enter if you want to play a game, yes or no").lower()
        if play_again in ("yes","no"):
            break
        
        else:
            print("Please answer only yes or no")
    if play_again == "no":
        print("Thank you for playing")
        break
```

## Word Guessing Game (Mini Hangman)
Objective:
Create a word-guessing game where the player tries to guess the letters of a hidden word.

This will test your:

●	Loops

●	Conditional logic

●	String manipulation

●	Exception handling

●	Logical thinking

Instructions:

1.	Create a list of random words (like “python”, “machine”, “learning”, “bridge”).

2.	Randomly select one word for the game.

3.	Show the word as blanks (_ _ _ _) and let the player guess one letter at a time.

4.	Keep track of wrong guesses (max 6 attempts).

5.	End the game when the player either guesses the full word or uses all chances


```python
import random

def hangman_guess():
    words=["books","python","datescientist","aiengineer"]
    
    secret_words=random.choice(words)
    
    
    attempts_left=6
    guessed_letters=[]
    
    hidden_word=["_" for _ in secret_words]
    
    print("Welcome to Word guessing game")
    print(" ".join(hidden_word))
    
    while attempts_left > 0 and "_" in hidden_word:
        try:
            user_guess=input("Enter the letter").lower()
            if len(user_guess) != 1 or not user_guess.isalpha():
                print("Please Enter the single letter")
            if user_guess in guessed_letters:
                print("You Already guessed that letter")
                continue
                
            guessed_letters.append(user_guess)
            
            if user_guess in secret_words:
                print("Correct guess")
                for i,kk in enumerate(secret_words):
                    if kk == user_guess:
                        hidden_word[i] = user_guess
            else:
                attempts_left -= 1
                print(f" Wrong guess ! Attempts_left : { attempts_left}")
        
            print("Words:"," ".join(hidden_word))
            print("Guessed letters:",guessed_letters)
    
        except ValueError as e:
            print(e)
    
    if "_" not in hidden_word:
        print("Congratulations! You guessed the word:",secret_words)
    else:
        print("Game Over! The word was:",secret_words)
        
```
